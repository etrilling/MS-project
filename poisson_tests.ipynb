```python
import numpy as np
from scipy.special import factorial
import matplotlib.pyplot as plt

def generate_poisson_events(rate, time_duration):
    num_events = np.random.poisson(rate * time_duration)
    event_times = np.sort(np.random.uniform(0, time_duration, num_events))
    inter_arrival_times = np.diff(event_times)
    return num_events, event_times, inter_arrival_times

num_events, event_times, inter_arrival_times = generate_poisson_events(1, 10000)


plt.hist(inter_arrival_times, bins=100, density=True)
# plot a poisson pdf with the same mean
mean = np.mean(inter_arrival_times)
x = np.linspace(0, 10, 1000)
# y = np.exp(-mean) * mean**x / factorial(x)
# plot a exponential pdf with the same mean
y = mean * np.exp(-mean * x)
plt.plot(x, y, 'r')
plt.show()
```

```python
# plot a sample of a normal distribution along with a normal pdf
sample = np.random.normal(0, 1, 10000)
plt.hist(sample, bins=100, density=True)
x = np.linspace(-4, 4, 1000)
y = 1 / np.sqrt(2 * np.pi) * np.exp(-x**2 / 2)
plt.plot(x, y, 'r')
plt.show()
```
