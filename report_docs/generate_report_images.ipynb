```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

y_noisy = y + 0.05 * np.random.normal(size=x.shape)

y_prime = np.cos(x)
y_finite_diff = np.diff(y) / np.diff(x)
y_noisy_finite_diff = np.diff(y_noisy) / np.diff(x)


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.plot(x, y, label='y = sin(t)')
ax1.plot(x, y_noisy, label='y = sin(t) + noise')
ax1.legend()

ax2.plot(x[1:], y_prime[1:], label='y\' = cos(t)')
ax2.plot(x[1:], y_finite_diff, '--', label='y\' = finite_diff(sin(t))')
ax2.plot(x[1:], y_noisy_finite_diff, linewidth=1, label='y\' = finite_diff(sin(t) + noise)')
# ax2.plot(x[1:], np.gradient(y_noisy[1:], x[1:]), linewidth=1, label='y\' = gradient(sin(x) + noise)')
ax2.legend()

plt.tight_layout()

plt.savefig('images/finite_diff_with_noise.png')
```
