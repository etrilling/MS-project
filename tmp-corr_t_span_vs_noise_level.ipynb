```python
dt = 0.01

noise_levels = np.logspace(-3, 0, 10).tolist()
t_spans = sorted([(0, 3*i) for i in range(1, 10 + 1)], reverse=True)

test_suite_config = {
    'lotka_volterra': {
        't_eval_test': np.arange(0, 50 + 0.01, 0.01),
        'noise_level': noise_levels,
        'dt_train': dt,
        't_span_train': t_spans
    },

    'van_der_pol_oscillator': {
        't_eval_test': np.arange(0, 25 + 0.01, 0.01),
        'noise_level': noise_levels,
        'dt_train': dt,
        't_span_train': t_spans
    },

    'duffing_oscillator': {
        't_eval_test': np.arange(0, 50 + 0.01, 0.01),
        'noise_level': noise_levels,
        'dt_train': dt,
        't_span_train': t_spans
    },

    'nonlinear_pendulum': {
        't_eval_test': np.arange(0, 25 + 0.01, 0.01),
        'noise_level': noise_levels,
        'dt_train': dt,
        't_span_train': t_spans
    },

    'lorenz': {
        't_eval_test': np.arange(0, 10 + 0.01, 0.01),
        'noise_level': noise_levels,
        'dt_train': 0.001,
        't_span_train': sorted([(0, 2*i) for i in range(1, 5 + 1)], reverse=True)
    },
}

list_of_results, list_of_experiment_configs = run_test_suite(test_suite_config, use_ESINDy=True, generate_prediction=False)
```

```python
def find_correct_system_points(experiment_config, results):
    list1_name, list1, list2_name, list2 = extract_list_items(experiment_config)
    
    # extract the true coefficients from the experiment configuration
    true_coeffs = experiment_config['true_coeffs']

    correct_system_pts = []
    for i in range(len(list1)):
        for j in range(len(list2)):
            model_xi = results[i][j]["model_xi"]

            # if the predicted model has the same non-zero coefficients as the true model, add this point
            if np.array_equal(np.nonzero(model_xi), np.nonzero(true_coeffs)):
                correct_system_pts.append([list1[i], list2[j]])
    
    return correct_system_pts    


# def calculate_correlation(experiment_config, results):
#     pts = find_correct_system_points(experiment_config, results)

#     pts = [(noise_level, end - start) for noise_level, (start, end) in pts]

#     pts = np.array(pts)
#     pts_x = pts[:, 1]
#     pts_y = pts[:, 0]

#     return np.corrcoef(pts_x, pts_y)[0, 1]


for system_name, experiment_config, results in zip(test_suite_config.keys(), list_of_experiment_configs, list_of_results):
    # cor = calculate_correlation(experiment_config, results)

    pts = find_correct_system_points(experiment_config, results)

    pts = [(noise_level, end - start) for noise_level, (start, end) in pts]

    if len(pts) == 0:
        continue

    pts = np.array(pts)
    pts_x = pts[:, 1]
    pts_y = pts[:, 0]

    a, b = np.polyfit(pts_x, pts_y, 1)
    cor = np.corrcoef(pts_x, pts_y)[0, 1]
    print(f'Correlation for {system_name} is {cor}, slope is {a}')

    plt.plot(pts_x, pts_y, 'o')
    plt.plot(pts_x, a * pts_x + b)
    # plt.gca().invert_xaxis()
    plt.show()
```
